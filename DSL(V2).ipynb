Lexer (lex)

```python
import ply.lex as lex
import sys

tokens = (
    'ID', 'STRING', 'NUMBER',
    'LBRACE', 'RBRACE', 'SEMICOL',
    'EQUAL', 'ARROW',
)

t_LBRACE = r'\{'
t_RBRACE = r'\}'
t_SEMICOL = r';'
t_EQUAL = r'='
t_ARROW = r'->'

t_ignore = ' \t'

def t_STRING(t):
    r'"[^"]*"'
    return t

def t_NUMBER(t):
    r'\d+'
    t.value = int(t.value)
    return t

def t_ID(t):
    r'[a-zA-Z_][a-zA-Z0-9_]*'
    return t

def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    print(f"Erro léxico: {t.value[0]}")
    t.lexer.skip(1)

# Workaround for ply in interactive environments like Colab
# Ensures __file__ is available in globals for ply's internal introspection
if '__file__' not in globals():
    globals()['__file__'] = 'lexer_script.py' # Provide a dummy file name

lexer = lex.lex(object=globals(), debug=False)
```

Parser (yacc)

```python
import ply.yacc as yacc

# Estrutura AST simples
class Scene:
    def __init__(self, name, commands):
        self.name = name
        self.commands = commands

class Say:
    def __init__(self, text):
        self.text = text

# Gramática

def p_program(p):
    '''program : program declaration
               | declaration'''
    if len(p) == 3:
        p[0] = p[1] + [p[2]]
    else:
        p[0] = [p[1]]

def p_declaration_scene(p):
    '''declaration : ID ID LBRACE commands RBRACE'''
    if p[1] == 'scene':
        p[0] = Scene(p[2], p[4])

def p_commands(p):
    '''commands : commands command
                | command'''
    if len(p) == 3:
        p[0] = p[1] + [p[2]]
    else:
        p[0] = [p[1]]

def p_command_say(p):
    '''command : ID STRING SEMICOL'''
    if p[1] == 'say':
        p[0] = Say(p[2])

def p_error(p):
    print("Erro sintático")

# ply.yacc.yacc does not accept an 'object' argument, only 'module'
parser = yacc.yacc(debug=False)
```

Teste

```python
code = '''
scene inicio {
    say "Olá mundo!";
}
'''

result = parser.parse(code)
print(result)
```
